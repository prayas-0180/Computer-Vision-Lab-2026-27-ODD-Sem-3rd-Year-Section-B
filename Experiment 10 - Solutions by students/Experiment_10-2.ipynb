```python
# Prayas Sharma
# CU24250180
# Sec - B
## ============================================================
# CNN-BASED IMAGE CLASSIFICATION USING MNIST
# Complete Experiment - Single Google Colab Cell
# ============================================================

# Install required libraries
!pip install -q tensorflow scikit-learn matplotlib seaborn

# ------------------------------------------------------------
# 1. Import Libraries
# ------------------------------------------------------------

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist
from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

print("=" * 65)
print("CNN-BASED HANDWRITTEN DIGIT CLASSIFICATION USING MNIST")
print("=" * 65)

# ------------------------------------------------------------
# 2. Load MNIST Dataset
# ------------------------------------------------------------

print("\n[1] Loading MNIST dataset...")

(X_train, y_train), (X_test, y_test) = mnist.load_data()

print("Training images :", X_train.shape)
print("Testing images  :", X_test.shape)
print("Number of classes:", len(np.unique(y_train)))

# ------------------------------------------------------------
# 3. Preprocessing
# ------------------------------------------------------------

print("\n[2] Preprocessing images...")

# Normalize pixel values from 0-255 to 0-1
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Add channel dimension
X_train = X_train.reshape(-1, 28, 28, 1)
X_test = X_test.reshape(-1, 28, 28, 1)

print("Preprocessed training shape:", X_train.shape)
print("Preprocessed testing shape :", X_test.shape)

# ------------------------------------------------------------
# 4. Display Sample Images
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_train[i].squeeze(), cmap="gray")
    plt.title(f"Label: {y_train[i]}")
    plt.axis("off")

plt.suptitle("Sample MNIST Images")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 5. Build CNN Model
# ------------------------------------------------------------

print("\n[3] Building CNN model...")

model = models.Sequential([

    # Convolutional Layer
    layers.Conv2D(
        32,
        (3, 3),
        activation="relu",
        input_shape=(28, 28, 1)
    ),

    # Pooling Layer
    layers.MaxPooling2D((2, 2)),

    # Second Convolutional Layer
    layers.Conv2D(
        64,
        (3, 3),
        activation="relu"
    ),

    # Second Pooling Layer
    layers.MaxPooling2D((2, 2)),

    # Flatten
    layers.Flatten(),

    # Fully Connected Layer
    layers.Dense(128, activation="relu"),

    # Output Layer
    layers.Dense(10, activation="softmax")
])

# ------------------------------------------------------------
# 6. Display Model Architecture
# ------------------------------------------------------------

model.summary()

# ------------------------------------------------------------
# 7. Compile CNN
# ------------------------------------------------------------

print("\n[4] Compiling model...")

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# ------------------------------------------------------------
# 8. Train CNN Model
# ------------------------------------------------------------

print("\n[5] Training CNN model...")

history = model.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=128,
    validation_split=0.1,
    verbose=1
)

print("\n✓ Training completed!")

# ------------------------------------------------------------
# 9. Plot Training and Validation Accuracy
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["accuracy"],
    label="Training Accuracy"
)

plt.plot(
    history.history["val_accuracy"],
    label="Validation Accuracy"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy")
plt.title("Training and Validation Accuracy")
plt.legend()
plt.grid(True)
plt.show()

# ------------------------------------------------------------
# 10. Plot Training and Validation Loss
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training and Validation Loss")
plt.legend()
plt.grid(True)
plt.show()

# ------------------------------------------------------------
# 11. Evaluate Model on Test Dataset
# ------------------------------------------------------------

print("\n[6] Evaluating CNN on test dataset...")

test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test,
    verbose=0
)

print(f"Test Loss     : {test_loss:.4f}")
print(f"Test Accuracy : {test_accuracy:.4f}")
print(f"Test Accuracy : {test_accuracy * 100:.2f}%")

# ------------------------------------------------------------
# 12. Generate Predictions
# ------------------------------------------------------------

print("\n[7] Generating predictions...")

y_probability = model.predict(X_test, verbose=0)

y_pred = np.argmax(
    y_probability,
    axis=1
)

# ------------------------------------------------------------
# 13. Calculate Evaluation Metrics
# ------------------------------------------------------------

accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

print("\n[8] PERFORMANCE METRICS")
print("-" * 50)

print(f"Accuracy  : {accuracy:.4f} ({accuracy * 100:.2f}%)")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1-Score  : {f1:.4f}")

# ------------------------------------------------------------
# 14. Classification Report
# ------------------------------------------------------------

print("\n[9] CLASSIFICATION REPORT")
print("-" * 65)

print(
    classification_report(
        y_test,
        y_pred,
        digits=4
    )
)

# ------------------------------------------------------------
# 15. Confusion Matrix
# ------------------------------------------------------------

print("\n[10] CONFUSION MATRIX")

cm = confusion_matrix(
    y_test,
    y_pred
)

plt.figure(figsize=(9, 7))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=range(10),
    yticklabels=range(10)
)

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("MNIST CNN Confusion Matrix")
plt.show()

# ------------------------------------------------------------
# 16. Display Predictions on Unseen Test Images
# ------------------------------------------------------------

print("\n[11] SAMPLE PREDICTIONS")

plt.figure(figsize=(12, 8))

for i in range(20):

    plt.subplot(4, 5, i + 1)

    plt.imshow(
        X_test[i].squeeze(),
        cmap="gray"
    )

    actual = y_test[i]
    predicted = y_pred[i]

    if actual == predicted:
        title = f"Actual: {actual}\nPredicted: {predicted}"
    else:
        title = f"Actual: {actual}\nPredicted: {predicted}"

    plt.title(title)
    plt.axis("off")

plt.suptitle("CNN Predictions on Unseen Test Images")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 17. Display Misclassified Images
# ------------------------------------------------------------

wrong_indices = np.where(
    y_test != y_pred
)[0]

print(
    f"\nNumber of misclassified images: "
    f"{len(wrong_indices)}"
)

plt.figure(figsize=(12, 8))

for i in range(min(20, len(wrong_indices))):

    index = wrong_indices[i]

    plt.subplot(4, 5, i + 1)

    plt.imshow(
        X_test[index].squeeze(),
        cmap="gray"
    )

    plt.title(
        f"Actual: {y_test[index]}\n"
        f"Predicted: {y_pred[index]}"
    )

    plt.axis("off")

plt.suptitle("Examples of Misclassified Images")
plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# 18. Final Summary
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("EXPERIMENT SUMMARY")
print("=" * 65)

print(f"Dataset              : MNIST")
print(f"Image Size           : 28 × 28")
print(f"Number of Classes    : 10")
print(f"Training Images      : {len(X_train)}")
print(f"Testing Images       : {len(X_test)}")
print(f"Epochs               : 5")
print(f"Optimizer            : Adam")
print(f"Test Accuracy        : {accuracy * 100:.2f}%")
print(f"Precision            : {precision:.4f}")
print(f"Recall               : {recall:.4f}")
print(f"F1-Score             : {f1:.4f}")
print(f"Misclassified Images : {len(wrong_indices)}")

print("\n✓ CNN image classification experiment completed successfully!")
```

1. What is image classification? How does it differ from object detection and image segmentation?

Image classification assigns a single class label to an image.
Object detection identifies objects and their locations using bounding boxes.
Image segmentation assigns a class label to each pixel of an image.

2. Explain the architecture and working principle of a CNN.

A CNN consists of Convolution, ReLU, Pooling, Flatten, Fully Connected, and Output layers. Convolution extracts image features, pooling reduces dimensions, and fully connected layers perform classification. The output layer gives the final class prediction.

⸻

3. What is the role of Convolutional, Pooling, and Fully Connected Layers?

* Convolution: Extracts features such as edges, textures, and shapes.
* Pooling: Reduces feature-map size and computation.
* Fully Connected: Uses extracted features to classify the image.

⸻

4. Why is image normalization performed?

Normalization scales pixel values, usually from 0–255 to 0–1. It improves numerical stability, speeds up training, and helps the model converge more effectively.

⸻

5. Explain ReLU and Softmax.

ReLU introduces non-linearity and is defined as max(0,x).
Softmax converts the final outputs into probabilities for different classes and selects the class with the highest probability.

⸻

6. What is a Confusion Matrix?

A Confusion Matrix is a table comparing actual and predicted classes. The diagonal shows correct predictions, while off-diagonal values show misclassifications. It helps analyze class-wise performance.

⸻

7. Differentiate Accuracy, Precision, Recall, and F1-Score.

* Accuracy: Percentage of total predictions that are correct.
* Precision: Percentage of predicted positives that are actually positive.
* Recall: Percentage of actual positives correctly identified.
* F1-Score: Harmonic mean of Precision and Recall.

⸻

8. Compare SIFT/HOG + SVM with CNN.

SIFT/HOG + SVM uses manually designed features and a separate classifier. CNN automatically learns features and performs classification together. CNNs generally handle complex image patterns better but require more data and computation.

10. How do Data Augmentation, Transfer Learning, and Hyperparameter Tuning improve performance?

* Data Augmentation: Creates varied training images and improves generalization.
* Transfer Learning: Reuses knowledge from a pre-trained model and reduces training requirements.
* Hyperparameter Tuning: Finds suitable settings such as learning rate, batch size, and epochs to improve performance.
